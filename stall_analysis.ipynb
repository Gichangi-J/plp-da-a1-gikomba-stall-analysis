```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions":   {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma":   {"qty": 9,  "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado":  {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1 – Daily Sales Analysis

I use Python's built-in `sum()`, `max()`, `min()` and `list.index()` functions to calculate the overall sales metrics and locate the best and worst sales days. The position in the 14-day list identifies the weekday and whether the result belongs to week 1 or week 2.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_week = best_index // 7 + 1
worst_week = worst_index // 7 + 1

print(f"Total sales: KES {total_sales:,.2f}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,.2f} on {days[best_index]}, Week {best_week}")
print(f"Lowest sales: KES {lowest_sales:,.2f} on {days[worst_index]}, Week {worst_week}")
```

## Task 2 – Weekly Sales and Growth

I define `week_totals()` to split the 14 daily values into two seven-day slices and return their totals as a tuple. I then compare week 2 with week 1 to calculate the percentage change.

```python
def week_totals(sales):
    """Return the sales totals for week 1 and week 2 as a tuple."""
    first_week_total = sum(sales[:7])
    second_week_total = sum(sales[7:14])
    return first_week_total, second_week_total


week_1_total, week_2_total = week_totals(daily_sales)
growth_percentage = ((week_2_total - week_1_total) / week_1_total) * 100

print(f"Week 1 total: KES {week_1_total:,.2f}")
print(f"Week 2 total: KES {week_2_total:,.2f}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

## Task 3 – Sales Day Classification

I classify each day's sales against the overall daily average. A day is Strong when it is more than 20% above average, Weak when it is more than 20% below average, and Normal otherwise. A dictionary stores the count for each category.

```python
def classify_day(amount, average):
    """Classify sales as Strong, Normal, or Weak relative to the average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


classification_counts = {"Strong": 0, "Normal": 0, "Weak": 0}
daily_classifications = []

for day_name, amount in zip(days, daily_sales):
    category = classify_day(amount, average_sales)
    daily_classifications.append((day_name, amount, category))
    classification_counts[category] += 1

for day_name, amount, category in daily_classifications:
    print(f"{day_name}: KES {amount:,.2f} – {category}")

print("\\nClassification totals:")
for category, count in classification_counts.items():
    print(f"{category}: {count} day(s)")
```

## Task 4 – Weekday vs Weekend Analysis

I use the weekday positions (Monday to Friday) and weekend positions (Saturday and Sunday) from each seven-day week. I calculate each group's average daily sales, then express how much higher or lower the weekend average is relative to the weekday average.

```python
weekday_sales = daily_sales[0:5] + daily_sales[7:12]
weekend_sales = daily_sales[5:7] + daily_sales[12:14]

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)
weekend_difference_percentage = ((weekend_average - weekday_average) / weekday_average) * 100

print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend average is {weekend_difference_percentage:.2f}% higher than the weekday average.")
```

## Task 5 – Stock Reordering and Gross Margins

I loop through the stock dictionary with `.items()` and select products whose current quantity is at or below their reorder level. For every product, I calculate gross margin as `(price - cost) / price × 100`.

```python
def products_to_reorder(inventory):
    """Return product names with stock at or below their reorder levels."""
    reorder_list = []
    for product, details in inventory.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)
    return reorder_list


reorder_list = products_to_reorder(stock)

print("Products to reorder:")
for product in reorder_list:
    print(f"- {product}: {stock[product]['qty']} kg in stock (reorder level: {stock[product]['reorder']} kg)")

print("\\nGross margins:")
gross_margins = {}
for product, details in stock.items():
    margin = ((details["price"] - details["cost"]) / details["price"]) * 100
    gross_margins[product] = margin
    print(f"{product}: {margin:.2f}%")
```

## Task 6 – Stall Performance Report

I combine the metrics calculated in the earlier tasks into one readable report. F-strings with `:,` format KES amounts with thousands separators, while percentages are displayed to two decimal places.

```python
print("=" * 54)
print("             STALL PERFORMANCE REPORT")
print("=" * 54)
print(f"Total sales (14 days):       KES {total_sales:,.2f}")
print(f"Average daily sales:         KES {average_sales:,.2f}")
print(f"Highest sales:               KES {highest_sales:,.2f} ({days[best_index]}, Week {best_week})")
print(f"Lowest sales:                KES {lowest_sales:,.2f} ({days[worst_index]}, Week {worst_week})")
print("-" * 54)
print(f"Week 1 sales:                KES {week_1_total:,.2f}")
print(f"Week 2 sales:                KES {week_2_total:,.2f}")
print(f"Week-on-week growth:         {growth_percentage:.2f}%")
print("-" * 54)
print(f"Weekday average:             KES {weekday_average:,.2f}")
print(f"Weekend average:             KES {weekend_average:,.2f}")
print(f"Weekend advantage:           {weekend_difference_percentage:.2f}%")
print("-" * 54)
print("DAY CLASSIFICATIONS")
for category in ("Strong", "Normal", "Weak"):
    print(f"{category:<12}: {classification_counts[category]} day(s)")
print("-" * 54)
print("PRODUCTS TO REORDER")
for product in reorder_list:
    print(f"{product}: {stock[product]['qty']} kg (reorder at {stock[product]['reorder']} kg)")
print("-" * 54)
print("GROSS MARGINS")
for product, margin in gross_margins.items():
    print(f"{product:<12}: {margin:.2f}%")
print("=" * 54)
```
